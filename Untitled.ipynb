```python
from core import RecurrentNetwork
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

brain = RecurrentNetwork(num_neurons=20,oscillator_frequency=100,timestep=0.01, change_rate=0.002)

heat_map = brain.state.numpy()

for i in range(1000):
    state = brain.state.numpy()
    if (i % 10 == 0):
        heat_map = np.concatenate((heat_map, state), axis=1)
    brain.step()
last = np.zeros(20).reshape(-1,1)
heat_map = np.concatenate((heat_map, last), axis=1)
print(heat_map[:, 0])
print(heat_map[:, 5])
fig, ax = plt.subplots()
state = brain.state.numpy()
im = ax.imshow(heat_map)
plt.show()
```
